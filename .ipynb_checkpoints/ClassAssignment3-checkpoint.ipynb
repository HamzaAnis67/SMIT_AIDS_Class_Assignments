# Class Assignment 15/11/26

```python
class Car:
    def __init__(self,Brand,Model,Engine,SeatCapacity,Color):
        self.Brand = Brand
        self.Model = Model
        self.Engine = Engine
        self.SeatCapacity = SeatCapacity
        self.Color = Color
    def Start(self):
        self.Start = "Car has started"
    def Accelarate(self,Speed):
        self.Speed = Speed
        self.Move = f"Car is moving with a speed of {self.Speed}"
    def Brake(self):
        self.Speed -= 10 
        self.Stop = f"Car decreases speed to {self.Speed}"
    def Display(self):
        print(f"""Car Attributes
Brand: {self.Brand}
Model: {self.Model}
Engine: {self.Engine}
Seat Capacity: {self.SeatCapacity}
Color: {self.Color}

Car Behavior's
The {self.Start}
The {self.Move}M/H
Due to traffic ahead {self.Stop}M/H
""")
```

```python
car1 = Car("Ford Mustang","2026","Supercharged 5.2L V8 (Predator)","4 seats","Red and Black")
```

```python
car1.Start()
```

```python
car1.Accelarate(50)
```

```python
car1.Brake()
car1.Brake()
```

```python
car1.Display()
```

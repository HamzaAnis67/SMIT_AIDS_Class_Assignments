```python
class MobileBattery:
    def __init__(self,mobile,battery):
        self.mobile = mobile 
        self._battery = battery
    def get_battery(self):
        print("The mobile battery percentage is:", self._battery)
    def set_battery(self,charge):
        if charge > 100 or charge < 0:
            print("Invalid Battery Percentage")
        else:
            self.NewCharge = self._battery + charge
            if self.NewCharge > 100 or self.NewCharge < 0:
                print("Invalid battery percentage, overcharged or undercharged")
            else:
                self._battery += charge
                print("Mobile battery charge updated to:",self._battery)
```

```python
h1 = MobileBattery("Redmi",65)
```

```python
h1.mobile
```

```python
h1.get_battery()
```

```python
h1.set_battery(101)
```

```python
h1.set_battery(55)
```

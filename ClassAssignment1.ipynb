Write a Python program using functions to develop a University Student Management and Result Evaluation System. The program should collect student information, evaluate academic performance, and generate a complete result report.
Requirements

1. Student Information: Take the student's name, age, roll number, and department as input from the user outside the functions.

2. Subject Marks: Ask the user to enter marks for five subjects and store them in a list.
3. Function with Parameters: Create a function named calculate_result() that accepts the student's information and subject marks as parameters.

4. Total Marks: Calculate the total marks obtained using a built-in function.

5. Percentage: Calculate the student's percentage. Each subject carries 100 marks.

6. Highest and Lowest Marks: Find the highest and lowest subject marks using built-in functions.

7. Pass or Fail: Use if-else statements to determine the result. The student must score at least 40 marks in every subject to pass.

8. Grade Evaluation: Assign a grade according to the following criteria:
   - 90% or above: A+
   - 80% to below 90%: A
   - 70% to below 80%: B
   - 60% to below 70%: C
   - 50% to below 60%: D
   - Below 50%: F
   If the student fails any subject, the final result must be Fail, regardless of the overall percentage.

9. Subject-Wise Report: Use a loop to display each subject's marks along with its subject number.

10. Dictionary: Store the student's personal information and academic results in a dictionary.

11. Return Statement: Return the calculated results from the function and store them in variables outside the function.

12. Formatted Report: Display a complete report using f-strings, including personal information, marks, total, percentage, highest marks, lowest marks, grade, and final result.

13. Department Message: Use another function named department_message() that accepts the department name as an argument and returns a suitable message for the student's department.

14. Final Display: Call both functions and display all information in a properly formatted report.

FINAL OUTPUT :

Student Name: Ali
Age: 20
Roll Number: AI-101
Department: Artificial Intelligence

Subject Marks:
Subject 1: 85
Subject 2: 92

Subject 4: 65
Subject 5: 80

Total Marks: 400 / 500
Percentage: 80.0%
Highest Marks: 92

Grade: A
Final Result: Pass

Department Message:

Welcome to the Artificial Intelligence Department.

```python
marks = []
studentName = input("Enter your full name:")
studentAge = int(input("Enter your age:"))
studentRollNo = input("Enter your Roll No (eg:AI-XX):")
studentDept = input("Enter your department full name:")
for i in range(5):
    mark = int(input(f"Enter marks obtained in subject {i+1}:"))
    marks.append(mark)
def calculate_result(name,age,rollno,dept,marks):
    totalMarks = sum(marks)
    percentage = 100 * (totalMarks / 500)
    lowestMarks = min(marks)
    highestMarks = max(marks)
    academicResult = {"Student Name": name,"Age": age,"Roll Number": rollno,"Department": dept,"Subject Marks": marks,"Total Marks": totalMarks,
                      "Percentage": percentage,"Highest Marks": highestMarks,"Lowest Marks": lowestMarks}
    if lowestMarks >= 40:
        if percentage >= 90:
            academicResult["grade"] = "A+"
            academicResult["result"] = "Pass"
        elif percentage >= 80:
            academicResult["grade"] = "A"
            academicResult["result"] = "Pass"
        elif percentage >= 70:
            academicResult["grade"] = "B"
            academicResult["result"] = "Pass"
        elif percentage >= 60:
            academicResult["grade"] = "C"
            academicResult["result"] = "Pass"
        elif percentage >= 50:
            academicResult["grade"] = "D"
            academicResult["result"] = "Pass"
        else:
            academicResult["grade"] = "F"
            academicResult["result"] = "Fail"
    else:
        academicResult["grade"] = "F"
        academicResult["result"] = "Fail"
    return academicResult
def department_message(department):
    return f"Welcome to the {department} department"

AcademicResult = calculate_result(studentName,studentAge,studentRollNo,studentDept,marks)
print(f"""
Student Name: {AcademicResult["Student Name"]}
Age: {AcademicResult["Age"]}
Roll Number: {AcademicResult["Roll Number"]}
Department: {AcademicResult["Department"]}
""")

print("Subject Marks:")
for i in range(len(marks)):
    print(f"Subject {i+1}: {marks[i]}")

print(f"""
Total Marks: {AcademicResult["Total Marks"]} / 500
Percentage: {AcademicResult["Percentage"]}%
Highest Marks: {AcademicResult["Highest Marks"]}
Lowest Marks: {AcademicResult["Lowest Marks"]}
Grade: {AcademicResult["grade"]}
Final Result: {AcademicResult["result"]}

Department Message:

{department_message(studentDept)}
""")
```

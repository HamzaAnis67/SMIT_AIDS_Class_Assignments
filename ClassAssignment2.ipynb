## Question:
: Employee Payroll and Performance Management System
Problem Statement:
Create a function named EmployeeManagement() that generates an employee record and calculates their monthly salary based on performance.
Required parameters:
EmployeeName
EmployeeID
Department
BasicSalary
WorkingDays
PresentDays
PerformanceRating
Additional information must be accepted through **ExtraDetails.
Requirements:
Store all required information in a dictionary.
Calculate attendance percentage.
Calculate salary after attendance deduction:
If attendance is 90% or above, no deduction.
If attendance is between 75% and 89%, deduct 5%.
If attendance is below 75%, deduct 15%.

Calculate performance bonus:
Rating 5: 20% bonus.
Rating 4: 15% bonus.
Rating 3: 10% bonus.
Below 3: No bonus.
Calculate final salary after deduction and bonus.
Add an employee performance category:
Rating 5: Excellent
Rating 4: Very Good
Rating 3: Good
Below 3: Needs Improvement
Add all additional details using update().
Return the complete employee dictionary.
Display the complete record using a loop.

```python
def EmployeeManagement(EmployeeName,EmployeeID,Department,BasicSalary,WorkingDays,PresentDays,PerformanceRating,**ExtraDetails):
    EmployeeRecord = {}
    EmployeeRecord["EmployeeName"] = EmployeeName
    EmployeeRecord["EmployeeID"] = EmployeeID
    EmployeeRecord["Department"] = Department
    EmployeeRecord["BasicSalary"] = BasicSalary
    EmployeeRecord["WorkingDays"] = WorkingDays
    EmployeeRecord["PresentDays"] = PresentDays
    EmployeeRecord["PerformanceRating"] = PerformanceRating
    EmployeeRecord.update(ExtraDetails)
    Attendance = (PresentDays / WorkingDays) * 100
    if Attendance >= 90:
        DeductedSalary = 0
    elif Attendance >= 75:
        DeductedSalary = (BasicSalary*0.05)
    else:
        DeductedSalary = (BasicSalary*0.15)
    if PerformanceRating == 5:
        PerformanceBonus = (BasicSalary*0.2)
        EmployeePerformance = "Excellent"
    elif PerformanceRating == 4:
        PerformanceBonus = (BasicSalary*0.15)
        EmployeePerformance = "Very Good"
    elif PerformanceRating == 3:
        PerformanceBonus = (BasicSalary*0.1)
        EmployeePerformance = "Good"
    else:
        PerformanceBonus = 0
        EmployeePerformance = "Needs Improvement"
    FinalSalary = BasicSalary - DeductedSalary + PerformanceBonus
    EmployeeRecord["DedutedSalary"] = DeductedSalary
    EmployeeRecord["PerformanceBonus"] = PerformanceBonus
    EmployeeRecord["FinalSalary"] = FinalSalary
    EmployeeRecord["EmployeePerformance"] = EmployeePerformance
    return EmployeeRecord
```

```python
EmployeeReport = EmployeeManagement("Muhammad Hamza Anis","067","Software Engineer",115000,22,20,5,Nationality= "Pakistan",Gender="Male")

print("==========Employee Report==========")
for key,values in EmployeeReport.items():
    print(f"{key} : {values}")
```
